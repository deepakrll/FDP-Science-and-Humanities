## 1. Setup and upload the matching CSV

```
!pip -q install seaborn scikit-learn opencv-python-headless
import io, warnings, numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from google.colab import files
from IPython.display import display
warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid",context="notebook")
plt.rcParams["figure.figsize"]=(9,5)
uploaded=files.upload()
csv_name=next(f for f in uploaded if f.lower().endswith(".csv"))
df=pd.read_csv(io.BytesIO(uploaded[csv_name]))
print("Loaded",csv_name,"shape:",df.shape)
display(df.head()); display(df.describe(include="all").T)
```

## 2. Calibration curve and unknown concentration estimation

```
from scipy.stats import linregress
cal=df.groupby("concentration_mg_L").absorbance_AU.agg(["mean","std"]).reset_index()
fit=linregress(df.concentration_mg_L,df.absorbance_AU)
xx=np.linspace(0,df.concentration_mg_L.max(),100)
plt.errorbar(cal.concentration_mg_L,cal["mean"],yerr=cal["std"],fmt="o",capsize=4,label="Mean ± SD")
plt.plot(xx,fit.intercept+fit.slope*xx,label=f"Linear fit R²={fit.rvalue**2:.4f}")
plt.xlabel("Concentration (mg/L)");plt.ylabel("Absorbance (AU)");plt.title("Beer–Lambert calibration");plt.legend();plt.show()
unknown_abs=0.52
print("Example unknown estimate (mg/L):",round((unknown_abs-fit.intercept)/fit.slope,2))
display(cal.round(4))
```
