## 1. Setup and upload the matching CSV

```
!pip -q install seaborn scikit-learn opencv-python-headless
import io, warnings, numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from google.colab import files
from IPython.display import display
warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid",context="notebook")
plt.rcParams["figure.figsize"]=(9,5)
uploaded=files.upload()
csv_name=next(f for f in uploaded if f.lower().endswith(".csv"))
df=pd.read_csv(io.BytesIO(uploaded[csv_name]))
print("Loaded",csv_name,"shape:",df.shape)
display(df.head()); display(df.describe(include="all").T)
```

## 2. Integrated rate-law plots and rate constant

```
from scipy.stats import linregress
fig,ax=plt.subplots(1,3,figsize=(15,4))
for c0,g in df.groupby("initial_concentration_mol_L"):
    g=g.sort_values("time_min")
    ax[0].plot(g.time_min,g.concentration_mol_L,"o-",label=f"C0={c0:.2f}")
    ax[1].plot(g.time_min,np.log(g.concentration_mol_L),"o-",label=f"C0={c0:.2f}")
    ax[2].plot(g.time_min,1/g.concentration_mol_L,"o-",label=f"C0={c0:.2f}")
for a,t in zip(ax,["C vs time","ln(C) vs time","1/C vs time"]):a.set_title(t);a.set_xlabel("Time (min)");a.legend()
plt.tight_layout();plt.show()
rows=[]
for c0,g in df.groupby("initial_concentration_mol_L"):
    fit=linregress(g.time_min,np.log(g.concentration_mol_L))
    rows.append([c0,-fit.slope,fit.rvalue**2,np.log(2)/(-fit.slope)])
display(pd.DataFrame(rows,columns=["initial_C_mol_L","k_per_min","R_squared","half_life_min"]).round(4))
```
