## 1. Setup and upload the matching CSV

```
!pip -q install seaborn scikit-learn opencv-python-headless
import io, warnings, numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from google.colab import files
from IPython.display import display
warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid",context="notebook")
plt.rcParams["figure.figsize"]=(9,5)
uploaded=files.upload()
csv_name=next(f for f in uploaded if f.lower().endswith(".csv"))
df=pd.read_csv(io.BytesIO(uploaded[csv_name]))
print("Loaded",csv_name,"shape:",df.shape)
display(df.head()); display(df.describe(include="all").T)
```

## 2. Estimate g and inspect model residuals

```
from scipy.stats import linregress
df["T_squared_s2"]=df.period_s**2
fit=linregress(df.length_m,df.T_squared_s2);xx=np.linspace(df.length_m.min(),df.length_m.max(),100)
plt.scatter(df.length_m,df.T_squared_s2,alpha=.6);plt.plot(xx,fit.intercept+fit.slope*xx,color="red")
plt.xlabel("Length (m)");plt.ylabel("T² (s²)");plt.title(f"Pendulum fit, R²={fit.rvalue**2:.4f}");plt.show()
g=4*np.pi**2/fit.slope
print("Estimated g (m/s²):",round(g,3),"| Percentage difference vs 9.81:",round(abs(g-9.81)/9.81*100,2),"%")
df["residual"]=df.T_squared_s2-(fit.intercept+fit.slope*df.length_m)
plt.scatter(df.length_m,df.residual);plt.axhline(0,ls="--",c="black");plt.xlabel("Length (m)");plt.ylabel("Residual (s²)");plt.title("Residual analysis");plt.show()
```
